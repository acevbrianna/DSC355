```python
#use previous data
#import libraries
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
#load data
df = pd.read_csv("eda_data.csv")

df.head()
```

```python
X = df.drop(columns=["y"])
y = df["y"]
```

```python
#check predictor
X.info()
```

Some predictor columns needed to be converted into a numerical format before they could be used in the regression model

```python
#split data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
#x train needs to be reviewed due to issues in code processing
print(X_train.dtypes)
print("\nMissing values:")
print(X_train.isnull().sum())
```

```python
#missing values review and use mean for inputs
X_train = X_train.fillna(X_train.mean(numeric_only=True))
X_test = X_test.fillna(X_train.mean(numeric_only=True))
```

```python
#text columns need to be addressed
X = pd.get_dummies(X, drop_first=True)
```

```python
#rerun
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
# Fill missing values FIRST
X_train = X_train.fillna(X_train.median(numeric_only=True))
X_test = X_test.fillna(X_train.median(numeric_only=True))

# Then create and train the model
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
# Check that missing values are gone
print("Missing values in X_train:", X_train.isnull().sum().sum())
print("Missing values in X_test:", X_test.isnull().sum().sum())
```

The dataset contained some missing values, which Linear Regression cannot process directly. I filled the missing predictor values with the median of each column so the model could be trained successfully.

```python
# Make predictions
y_pred = model.predict(X_test)

print(y_pred[:10])
```

After training the model, I used the testing data to generate predicted values for y.

```python
# Compare actual and predicted
results = pd.DataFrame({
    "Actual": y_test,
    "Predicted": y_pred
})

results.head(10)
```

I compared the actual values of y with the values predicted by the Linear Regression model to see how closely the predictions matched the real results.

I am going to calculate the R^2 score to get a basic idea of how well the Linear Regression model explained the variation in the target variable. A higher R^2 score means the model explains more of the variation in y.

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)

print(f"R² Score: {r2:.4f}")
```

Overall, I used Linear Regression to predict the continuous target variable y. Before training the model, I handled the missing values so the data could be used correctly. I split the dataset into training and testing groups, using 80% of the data for training and 20% for testing. After the model was trained, I used it to make predictions and compared those predictions to the actual values. I also calculated the R^2 score to get an idea of how well the model explained the changes in y.
